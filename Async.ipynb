```python
import time

def do_work(task_id: int, duration: float = 0.1) -> str:
    time.sleep(duration)
    return f"Task {task_id} completed"

def run_sync(tasks: int= 5) -> list[str]:
    results: list[str] = []

    for i in range(tasks):
        result = do_work(i, duration=3)
        results.append(result)
        print(result,'/n')
    return results

print(run_sync())
```
